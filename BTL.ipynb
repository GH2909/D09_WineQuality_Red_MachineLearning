```python
import pandas as pd
file_path = "/content/D09 - winequality-red.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="winequality-red"
)

print("===== 5 DÒNG ĐẦU CỦA DỮ LIỆU =====")
df.head()
```

```python
# Kích thước và chiều của dữ liệu
print("\n===== 1a. KÍCH THƯỚC VÀ CHIỀU DỮ LIỆU =====")
print("Số chiều (ndim):", df.ndim)
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])
print("Tổng số phần tử:", df.size)
```

```python
# Kiểu dữ liệu của các thuộc tính
print("\n===== 1b. KIỂU DỮ LIỆU CÁC THUỘC TÍNH =====")
print(df.dtypes)
```

```python
# Số lượng thực thể của các giá trị nhãn
print("\n===== 1c. PHÂN BỐ NHÃN QUALITY =====")
print(df["quality"].value_counts().sort_index())
print(df["quality"].value_counts().sum())
```

```python
#  Min, Max và Mean của các thuộc tính số
print("\n===== 1d. MIN - MAX - MEAN CÁC THUỘC TÍNH =====")

X = df.drop(columns=["quality"])

statistics = X.agg(["min", "max", "mean"]).T

print(statistics.round(3))
```

```python
# Kiểm tra dữ liệu thiếu
print("\n===== KIỂM TRA GIÁ TRỊ THIẾU =====")

print(df.isnull().sum())
```
